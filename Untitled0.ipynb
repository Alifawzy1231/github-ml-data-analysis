```python
import requests
import pandas as pd

url = "https://api.github.com/search/repositories?q=machine+learning&sort=stars&order=desc&per_page=100"

response = requests.get(url)

print(response.status_code)

data = response.json()
```

```python
df = pd.DataFrame(data["items"])

print(df.shape)
display(df.head())
```

```python
df = df[
    [
        "name",
        "owner",
        "language",
        "stargazers_count",
        "forks_count",
        "watchers_count",
        "open_issues_count",
        "created_at",
        "updated_at",
        "license"
    ]
]

display(df.head())
```

```python
df["owner"] = df["owner"].apply(
    lambda x: x["login"] if isinstance(x, dict) else None
)

df["license"] = df["license"].apply(
    lambda x: x["spdx_id"] if isinstance(x, dict) else None
)

display(df.head())
```

```python
print(df.isnull().sum())

df["language"] = df["language"].fillna("Unknown")
df["license"] = df["license"].fillna("No License")
df["owner"] = df["owner"].fillna("Unknown")
```

```python
print("Duplicates:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates after removal:", df.duplicated().sum())
```

```python
df["created_at"] = pd.to_datetime(df["created_at"])
df["updated_at"] = pd.to_datetime(df["updated_at"])

df = df.rename(columns={
    "stargazers_count": "stars",
    "forks_count": "forks",
    "watchers_count": "watchers",
    "open_issues_count": "open_issues",
    "created_at": "created_date",
    "updated_at": "updated_date"
})

display(df.head())
```

```python
df.to_csv("github_projects.csv", index=False)

print("github_projects.csv saved successfully!")

# Verify
df_check = pd.read_csv("github_projects.csv")

print(df_check.shape)
display(df_check.head())
```

```python
import sqlite3
import pandas as pd

# Load the CSV file
df = pd.read_csv("github_projects.csv")

# Create SQLite database
conn = sqlite3.connect("github_projects.db")

# Store the data in a table called Repositories
df.to_sql("Repositories", conn, if_exists="replace", index=False)

print("Database created successfully!")
print("Repositories table created successfully!")
```

```python
query = "SELECT * FROM Repositories LIMIT 5;"

result = pd.read_sql_query(query, conn)

display(result)
```

```python
# 1. Repositories with more than 10,000 stars

query_1 = """
SELECT name, owner, stars
FROM Repositories
WHERE stars > 10000
ORDER BY stars DESC;
"""

result_1 = pd.read_sql_query(query_1, conn)

print("Repositories with more than 10,000 stars:")
display(result_1)
```

```python
# 2. Repository names containing "Machine"

query_2 = """
SELECT name, owner, language, stars
FROM Repositories
WHERE name LIKE '%Machine%';
"""

result_2 = pd.read_sql_query(query_2, conn)

print("Repositories containing 'Machine':")
display(result_2)
```

```python
# AND
query_3 = """
SELECT name, language, stars, forks
FROM Repositories
WHERE stars > 1000 AND forks > 100;
"""

result_3 = pd.read_sql_query(query_3, conn)

print("AND query:")
display(result_3)
```

```python
# OR
query_4 = """
SELECT name, language, stars
FROM Repositories
WHERE language = 'Python' OR language = 'Jupyter Notebook';
"""

result_4 = pd.read_sql_query(query_4, conn)

print("OR query:")
display(result_4)
```

```python
# NOT
query_5 = """
SELECT name, language, stars
FROM Repositories
WHERE NOT language = 'Python';
"""

result_5 = pd.read_sql_query(query_5, conn)

print("NOT query:")
display(result_5)
```

```python
query_6 = """
SELECT name, owner, language, stars
FROM Repositories
ORDER BY stars DESC
LIMIT 10;
"""

top_10 = pd.read_sql_query(query_6, conn)

print("Top 10 repositories by stars:")
display(top_10)
```

```python
# Total number of repositories

query_7 = """
SELECT COUNT(*) AS total_repositories
FROM Repositories;
"""

total_repositories = pd.read_sql_query(query_7, conn)

print("Total repositories:")
display(total_repositories)
```

```python
# Average number of stars

query_8 = """
SELECT AVG(stars) AS average_stars
FROM Repositories;
"""

average_stars = pd.read_sql_query(query_8, conn)

print("Average stars:")
display(average_stars)
```

```python
query_9 = """
SELECT language, COUNT(*) AS repository_count
FROM Repositories
GROUP BY language
HAVING COUNT(*) > 5
ORDER BY repository_count DESC;
"""

language_groups = pd.read_sql_query(query_9, conn)

print("Languages with more than 5 repositories:")
display(language_groups)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.barh(top_10["name"], top_10["stars"])

plt.xlabel("Stars")
plt.ylabel("Repository")
plt.title("Top 10 Most Popular Machine Learning Repositories")

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
# Convert created_date to datetime
df["created_date"] = pd.to_datetime(df["created_date"])

# Count repositories by year
creation_trend = (
    df.groupby(df["created_date"].dt.year)
      .size()
      .reset_index(name="repository_count")
)

plt.figure(figsize=(10, 6))

plt.plot(
    creation_trend["created_date"],
    creation_trend["repository_count"],
    marker="o"
)

plt.xlabel("Year")
plt.ylabel("Number of Repositories")
plt.title("Repository Creation Trends Over Time")

plt.tight_layout()
plt.show()
```
